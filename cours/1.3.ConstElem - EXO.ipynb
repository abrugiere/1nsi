### Exercice 7

On part de 1 et tant que le nombre au carré est inférieur à 1000000000, on ajoute 1 au nombre.

````python
# On part de 1
n = 1

# Tant que n² < 1000000000 on passe au n suivant
# n = n +1

while n**2 < 1000000000:
    n = n + 1

# Quand on s'arrête c'est que n² > 1000000000
# Donc on affiche n
print("Fin de la boucle, donc n² >1000000000")

print(n)
````

### Exercice 10
Test de l'algorithme avec `n = 4` et `x = 3` :

| Boucle             | Valeur de `res` | Valeur de `cpt` |
|:------------------ |:---------------:| :--------------:|
| Initialisation     | 0               | 4               |
| While 1            | 0 + 3 = 3       | 4 - 1 = 3       |
| While 2            | 3 + 3 = 6       | 3 - 1 = 2       |
| While 3            | 6 + 3 = 9       | 2 - 1 = 1       |
| While 4            | 9 + 3 = 12      | 1 - 1 = 0       |

Fin de la boucle donc on affiche `res` qui vaut 12
La boucle a ajouté `n` "fois"  `x`, elle a donc fait le produit de `n` et `x`


````python
# Vérification

n = 4
x = 3

cpt = n
res = 0
while cpt > 0:
    res = res + x
    cpt = cpt - 1
print(res)
````

### Exercice 15
- Ici nous demandons un nombre de valeurs qui va donner le nombre d'éléments à saisir, donc le nombre d'itérations de la boucle `for` qui demandera les nombres.  
- Ensuite, on répète la demande des nombres à additionner avec la boucle et on accumule la somme (comme dans l'exo 12)
- A la fin on affichera la somme
- Pour la moyenne, il suffit de diviser la somme par le nombre d'éléments

````python
# demande le nombre de valeurs
nb_val = int(input('Combien de valeurs ? '))

# Initialisation de l'accumulateur
somme = 0

# Ici on n'utilisera pas le compteur
for _ in range(nb_val):
    # On demande la valeur
    nb = int(input('Valeur : '))
    # On ajoute la valeur à la somme
    somme = somme + nb

# On affiche la somme
print("Somme =",somme)

# On calcule et affiche la moyenne
moyenne = somme / nb_val
print("Moyenne =",moyenne)
````

- On écrit les nombres de 1 à 20 : on tulise une boucle `for` avec les bonnes bornes
- Ensuite pour chaque valeur, on utilise un `if` pour afficher le bon nombre 

````python
# Boucle de 1 à 20
for nombre in range(1,21):
    # Divisible par 3 et 5
    if nombre % 3 == 0 and nombre % 5 == 0:
        print("FizzBuzz")
    # Divisible par 3 uniquement
    elif nombre % 3 == 0:
        print("Fizz")
    # Divisible par 5 uniquement
    elif nombre % 5 == 0:
        print("Buzz")
    # Dernier cas
    else :
        print(nombre)
````
