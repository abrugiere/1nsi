### Spécification 

Nous avons abordé dans les cours précédents les **docstrings**, le **nommage des variables**, les **commentaires** éléments importants et nécessaires pour produire un code compréhensible et de qualité.  

Ces éléments sont appelés couramment la **spécification** des fonctions.  

Un autre élément important de la spécification des fonctions est le **prototypage** des fonctions : le fait de désigner le type des paramètres et des retours d'une fonction. Reprenons la fonction `maximum` des exercices précédents :

```python
# Définition de la fonction avec docstring
def maximum(n1, n2):
    """ Renvoie le maximum de n1 et n2. n1 et n2 sont deux entiers """
    if n1 > n2:
        return n1
    else:
        return n2
```

La docstring est alors disponible en utilisant la fonction `help` :

```python
help(maximum)
```

La fonction prend en paramètres deux entiers et renvoie un entier. Les types de ces paramètres pourront donc être précisés de la manière suivante :

```python
def maximum(n1 : int, n2 : int) -> int :
    """ Renvoie le maximum de n1 et n2. n1 et n2 sont deux entiers """
    if n1 > n2:
        return n1
    else:
        return n2
```

```python
help(maximum)
```

!!! tip Pour prototyper une fonction
- On ajoute le type de chaque paramètre à la suite de celui-ci suivi de deux points : `n1: int`
- On ajoute le type de la valeur de retour suivi d'une flèche : `-> int`
!!!

## Les assertions

### Le mot clé `assert`

À la manière du mot clé `return`, lorsqu'il est exécuté, mot clé `assert` quitte le programme en cours si il est suivi d'une expression boolénne dont la valeur est `False`, dans ce cas il lèvera une erreur 'AssertionError' sinon le programme continuera.

```python
# Exemple assert 1
assert 1 == 2
```

```python
# Exemple assert 2
assert 3 > 2
```

Il est possible de mettre une chaîne de caractères à la suite de l'assert pour l'afficher lors de l'erreur :

```python
# Exemple assert 3
assert 1 == 2, "Les nombres sont différents"
```

L'utilisation de ce mot clé est appelé une assertion :

!!! info Définition
Une **assertion** est une ligne de code qui vérifie une valeur booléenne, si cette valeur est fausse, le programme s'arrêtera et renverra une erreur.
!!!

### Vérification des pré-conditions
Une fonction demande, pour être correcte, d'utiliser des paramètres du bon type et dans un intervalle prévu par le développeur. C'est ce qu'on appelle les **pré-conditions**. L'utilisation de mauvais paramètres peut avoir des conséquences sur l'exécution de la fonction (*exemple de StackOverflow*).  

On utilise des **assertions** pour vérifier que les valeurs passées aux paramètres (les arguments) sont du bon type.  

Dans l'exemple ci dessous, vous pouvez constater que la fonction écrite ne vérifie pas si l'entier `b` passé en paramètre est égal à $0$ et plante :

```python
def divise(a : int, b : int) -> float :
    """ Renvoie la division de a par b, a et b sont 2 entiers """
    
    return a/b
```

```python
divise(1,0)
```

On peut détecter ceci en ajoutant une assertion à l'aide du mot clé `assert` :

```python
def divise(a,b) :
    """ Renvoie la division de a par b, a et b sont 2 entiers """
    
    # Pré-condition
    assert b!=0, 'Le second entier doit être non-nul'
    
    return a/b
```

```python
divise(1,0)
```

Evidemment, le programme plante encore, mais pouvons maintenant personnaliser le message d'erreur pour le rendre plus parlant.  

Les préconditions sont également souvent utilisées pour vérifier les types des paramètres :

```python
def divise(a,b) :
    """ Renvoie la division de a par b, a et b sont 2 entiers """
    
    # Pré-conditions
    assert b!=0, 'Le second entier doit être non-nul'
    assert type(a) == int, 'Les nombres saisis doivent être des entiers'
    assert type(b) == int, 'Les nombres saisis doivent être des entiers'

    return a/b
```

```python
divise(2.5,10)
```

```python
divise(1,'x')
```

### Tester ses fonctions

Les assertions peuvent également servir à tester votre programme, dans notre cas, on sait par exemple que `divise(6,3)` vaudra 2.  
On peut donc tester cette égalité avec une ou des assertions qu'on ajoutera à la suite de nos fonctions  :

```python
def divise(a,b) :
    """ Renvoie la division de a par b, a et b sont 2 entiers """
    
    # Pré-conditions
    assert b!=0, 'Le second entier doit être non-nul'
    assert type(a) == int, 'Les nombres saisis doivent être des entiers'
    assert type(b) == int, 'Les nombres saisis doivent être des entiers'

    return a/b

# Tests de la fonction
assert divise(6,3)==2, 'Erreur de codage'
```

Comme l'assertion est vraie, le programme ne renvoie pas d'erreur, mais si le programme ne renvoie pas le résultat attendu on pourra le voir avec l'assertion :

```python
def divise(a,b) :
    """ Renvoie la division de a par b, a et b sont 2 entiers """
    
    # Pré-conditions
    assert b!=0, 'Le second entier doit être non-nul'
    assert type(a) == int, 'Les nombres saisis doivent être des entiers'
    assert type(b) == int, 'Les nombres saisis doivent être des entiers'

    return a  # code erroné pour lever une erreur

# Tests de la fonction
assert divise(6,3)==2, 'Erreur de codage'
```

Il est également possible de créer des fonctions de test pour nos fonctions, et mettre dans celle-ci tous les cas possibles :

```python
def est_pair(n : int) -> bool:
    """ Renvoie True si le nombre est pair, False sinon"""

    # Précondition : type du paramètre : int
    assert type(n) == int, 'Le nombre saisi doit être un entier'
    
    if n%2 == 0 :
        return True
    else :
        return False

def test_est_pair():
    assert est_pair(8) == True, 'Problème avec un nombre pair'
    assert est_pair(7) == False, 'Problème avec un nombre impair'
    print('Tests OK')
```

```python
# Exécution de la fonction de tests
test_est_pair()
```

Tests avec une fonction mal codée :

```python
def est_pair(n : int) -> bool:
    """ Renvoie True si le nombre est pair, False sinon"""

    # Précondition : type du paramètre : int
    assert type(n) == int, 'Le nombre saisi doit être un entier'
    
    if n%2 == 0 :
        return True
    else :
        return True  # Code erroné pour lever une erreur

def test_est_pair():
    assert est_pair(8) == True, 'Problème avec un nombre pair'
    assert est_pair(7) == False, 'Problème avec un nombre impair'
    print('Tests OK')
```

```python
# Exécution de la fonction de tests
test_est_pair()
```
