## Introduction

Pendant la réalisation d'un programme informatique, nous écrivons des blocs de code qui permettent de produire un résultat.  
Dans le cadre d'un projet conséquent, on peut avoir besoin de **refaire** des opérations déjà effectuées précédemment.  
La solution la plus simple est de _copier/coller_ le code précédent. Cette méthode sera fonctionnelle, mais si vous devez modifier ce code, vous allez devoir le refaire partout où vous avez copié ce code, ceci peut être une source d'erreur en cas d'oubli. De plus votre code comportera de nombreuses lignes qui se répetent, ce qui rendra votre programme peu lisible.  

Les langages évolués permettent d'écrire des blocs de code, de leur donner un **nom**, et de **rappeler** ce code quand on en a besoin en utilisant son nom : **les fonctions**.  

Nous avons déjà utilisé des fonctions depuis le début de l'année : `print`, `input`, `range`...

!!! abstract Définir une fonction en Python
- Pour définir une fonction en Python, on utilise le mot clé `def` suivi du **nom** de la fonction puis de **parenthèses** avec des paramètres (nous allons aborder ce sujet plus loin) et enfin de deux points `:`
- Les instructions qui composeront la fonction appelées **le corps de la fonction** seront **indentées** sous l'instruction `def`.
!!!

---
## Exemple

Le programme suivant est issu de l'_exercice 12_ du chapitre _Constructions élémentaires_. Il calcule la somme des 1000 premiers entiers naturels et l'affiche à l'aide d'un `print` :

```python
# Calcul de la somme des 1000 premiers entiers naturels
s = 0
for k in range(1, 1001):
    s = s + k
print(s)
```

On peut donc réécrire ce programme sous la forme d'une fonction. Exécutez-le. Que remarquez vous ?

```python
# Définition de la fonction somme_1000
def somme_1000():
    s = 0
    for k in range(1, 1001):
        s = s + k
    print(s)
```

!!! bug Rien ne s'affiche !  

C'est normal, en exécutant ce code, vous avez dit à l'ordinateur  
>_Voici le code de la fonction nommée `somme_1000`_


L'interpréteur Python a donc **mémorisé** ce code et il est prêt à l'exécuter. 
!!!

Vous devez maintenant lui demander d'exécuter ce code. Pour cela, vous utiliserez son nom :

```python
# Exécution de la fonction somme_1000
somme_1000()
```

!!! warning Attention aux parenthèses
Vous noterez que lors de l'appel de la fonction, comme lors de sa définition, nous avons ajouté des parenthèses à la fin.  
Lors de l'appel de la fonction, comme lors de sa définition, les parenthèses sont **indispensables**.
!!!

```python
# Exécution sans les parenthèses
somme_1000
```

## Les paramètres

Nous disposons donc maintenant d'une fonction qui permet de calculer la somme des 1000 premiers entiers naturels.  
Si nous voulons maintenant calculer la somme des 2000 premiers entiers naturels, nous ne pourrons pas utiliser la fonction `somme_1000` dans l'état, mais nous pouvons nous en inspirer et écrire la fonction `somme_2000` :

```python
# Définition de la fonction somme_2000 à partir de somme_1000
def somme_2000():
    s = 0
    for k in range(1, 2001):
        s = s + k
    print(s)
```

```python
# Exécution de la fonction somme_2000
somme_2000()
```

Ça fonctionne ! Mais maintenant si je veux la somme des 3000 premiers entiers, et celle des 4000 premiers entiers ? Je vais devoir recopier la fonction encore et encore...  

En copiant la fonction `somme_1000` nous avons seulement changé 1 chiffre dans celle-ci, tout le reste du code est exactement le même.  

Python permet d'utiliser une variable `n` qui permettra de dire :  
>_Pour tout entier `n` la somme des `n` premiers entiers naturels est..._

Ceci s'écrira de la manière suivante :

```python
# Définition de la fonction somme
def somme(n):
    s = 0
    for k in range(1, n+1):
        s = s + k
    print(s)
```

Comme précédemment, la fonction est "chargée en mémoire" et prête à être exécutée.  

Maintenant, pour demander à l'interpréteur Python :
>_Calcule la somme pour l'entier `n = 3000`_

nous écrirons :

```python
# Appel de la fonction somme pour les 3000 premiers entiers
somme(3000)
```

!!! info Vocabulaire
- La valeur `n` est appelée **paramètre** de la fonction `somme`.
- On dit qu'on **passe** le paramètre `n` à la fonction  `somme`.
- Dans l'exemple ci-dessus, on dit qu'on a appelé la fonction  `somme` avec **l'argument** `3000`.
!!!

La variable `n` permet de représenter une valeur abstraite.  

En mathématiques, lorsqu'on écrit $f(x) = x^2 + 2x + 1$, la variable `x` représente un nombre quelconque et on aurait pu écrire $f(y) = y^2 + 2y + 1$ ou encore $f(n) = n^2 + 2n + 1$ pour représenter la même fonction.

Lorsqu'on voudra calculer par exemple l'image de $5$ par $f$, on remplacera `x` par $5$ et on écrira :
$$f(5) = 5^2 + 2\times5 + 1 = 36$$

De la même manière, on aurait pu écrire la fonction `somme` avec un autre nom de variable et obtenir le même résultat :

```python
# Définition de la fonction somme avec un paramètre de nom différent
def somme(nombre):
    s = 0
    for k in range(1, nombre+1):
        s = s + k
    print(s)
```

```python
# Appel de la fonction somme pour les 3000 premiers entiers
somme(3000)
```

!!! tip Nombre de paramètres d'une fonction
- Une fonction peut avoir **plusieurs paramètres**.
- Il suffit lors de la déclaration et de l'appel de la fonction de les **séparer par une virgule**.  

La fonction suivante permet de répéter `n` fois le mot `chaine` :
!!!

```python
# Définition de la fonction avec 2 paramètres
def repete(n, chaine):
    for i in range(n):
        print(chaine)
```

```python
# Appel de la fonction avec les arguments 5 et "Miaou"
repete(5,'Miaou')
```

!!! warning L'ordre des paramètres est important
Le code ci-dessous est incorrect
!!!

```python
# Appel de la fonction avec les arguments "Miaou" et 5
repete('Miaou',5)
```

## Les valeurs retournées

### Problématique

Les fonctions définies ci-dessus nous ont permis d'effectuer des opérations et **d'afficher** leur résultat.  

A présent, j'aimerais calculer la somme des 2000 premiers entiers puis calculer la somme des 1000 premiers entiers et enfin calculer le quotient de ces deux nombres.  

Pour cela, il me faudrait la possibiliter de stocker la valeur des calculs dans des variables pour pouvoir faire le quotient des deux variables.  

Essayons le code suivant :

```python
nombre_1 = somme(2000)
print("nombre 1 = ",nombre_1)
nombre_2 = somme(1000)
print("nombre 2 = ",nombre_2)

resultat = nombre_1 / nombre_2

print(resultat)
```

Vous constatez que les deux appels à la fonction `somme` ont fonctionné (affichage des résultats) mais les valeurs calculées ne sont pas enregistrées dans les variables `nombre_1` et  `nombre_2`.  
Ensuite, le calcul de mon résultat provoque une erreur.  

Il faudrait qu'au lieu d'afficher la valeur calculée, la fonction `somme` me la **renvoie** pour que je puisse la stocker dans les variables.

### Le mot clé `return`

Python permet à une fonction de **renvoyer** son résultat, c'est a dire que le résultat de l'appel de la fonction sera une expression qui aura une valeur d'un type particulier.  

Ainsi, nous pouvons ré-écrire la fonction `somme` pour qu'elle **retourne** (et non affiche) la valeur calculée, il suffira de remplacer l'appel de la fonction `print` par le mot clé `return` :

```python
# Définition de la fonction somme avec renvoi de la valeur par return
def somme(n):
    s = 0
    for k in range(1, n+1):
        s = s + k
    return s
```

```python
# Appel de la fonction 
somme(1000)
```

Vous constatez ici que le résultat est toujours bon, mais cette fois au lieu d'afficher la valeur, vous l'avez dans la sortie console (Affichage de `Sortie[xx]:` devant le nombre). Ceci est un fonctionnement de Jupyter Notebook, mais dans IDLE vous n'auriez rien d'affiché.

!!! warning Affichage de la valeur retournée par une fonction
Dans un IDE, si vous souhaitez afficher la valeur retournée par une fonction dans vos programmes, vous devrez utiliser un `print` combiné à l'appel de votre fonction.
!!!

Vous devrez donc plutôt écrire :

```python
# Appel de la fonction pour affichage de la valeur
print(somme(1000))
```

Testons maintenant notre code qui calcule le quotient :

```python
nombre_1 = somme(2000)
print("nombre 1 = ",nombre_1)
nombre_2 = somme(1000)
print("nombre 2 = ",nombre_2)

resultat = nombre_1 / nombre_2

print(resultat)
```

Tout fonctionne !  
Vous noterez que les résulats ne sont plus affichés, mais cette fois les variables `nombre_1` et  `nombre_2` ont bien pu être affectées avec les sommes calculées.

!!! info Vocabulaire
- La valeur après le mot clé `return` est appelée **retour** de la fonction.
- On dit que la fonction **retourne** la valeur `2001000` pour le premier appel.
!!!

!!! warning `print` ou `return` ?
Vous allez désormais arrêter d'utiliser `print` pour le résultat d'une fonction et utiliser plutôt le `return`. De manière générale :
- Si on vous demande d'écrire une fonction qui **retourne** ou **renvoie** quelque chose vous utiliserez impérativement un **`return`**
- Si on vous demande d'écrire une fonction qui **affiche** ou **liste** quelque chose vous pourrez utiliser **`print`**
!!!

### Un siège éjectable

Le comportement du mot clé `return` est particulier : lorsque la fonction exécute celui-ci, **elle s'arrête immédiatement** et renvoie la valeur derrière le mot clé.  

La fonction suivante :
- Affiche si le nombre passé en paramètre est pair,
- Renvoie la valeur du quotien de la division euclidienne de ce nombre par 2

Appelez-la avec différentes valeurs et observez son comportement :

```python
# Définition de la fonction parite
def parite(n):
    if n%2 == 0:
        print("Le nombre", n, "est pair")
        return n//2
    else:
        print("Le nombre", n, "est impair")
        return n//2

    print("FIN")
```

```python
# Test avec un nombre pair (16), quotient = 8
a = parite(16)
print(a)
```

```python
# Test avec un nombre impair (7), quotient = 3
b = parite(7)
print(b)
```

Vous notez ici, que l'instruction `print("FIN")` n'est jamais appelée.  
- Si le nombre est pair on affiche qu'il est pair et on renvoie le quotient donc on quitte la fonction
- Si le nombre est impair on affiche qu'il est impair et on renvoie le quotient donc on quitte également la fonction

Le code situé en dessous du `if` n'est donc jamais exécuté.


### Les procédures

!!! info Définition
Une fonction qui ne renvoie pas de valeur (qui n'utilise pas de `return`) est appelée une **procédure**.  
La valeur renvoyée par celle-ci sera **`None`** (qui représente **"RIEN"** en Python)
!!!

```python
# Définition de la fonction sans return
def miaule(n):
    print(n*"miaou ")
```

```python
# Appel et sauvegarde du résultat de la fonction dans une variable
a = miaule(12)
```

```python
# Vérification de la valeur de a
print(a)
```

## Variables et fonctions : espaces et portée des variables

!!! info Définition
- Dans un programme, les variables créées en dehors des fonctions sont appelées **variables globales**.
- Les variables crées dans une fonction ainsi que les paramètres de la fonction sont appelées **variables locales**.
!!!

On dit que les fonctions créent leur «espace de noms», un espace qui leur est propre.

!!! tip Règles d'accès en lecture et en modification d'une variable suivant son espace d'origine
- Règle 1 : Une variable locale est **inaccessible** hors de sa fonction.
- Règle 2 : Une variable globale est accessible en **lecture** à l'intérieur d'une fonction.
- Règle 3 : Une variable globale **ne peut pas être modifiée** à l'intérieur d'une fonction.
!!!


**Règle 1 :** Les variables locales n'existent **que dans votre fonction**, il ne sera pas possible de les utiliser en dehors de la fonction, même si vous avez déjà fait appel à celle-ci.  

```python
# Définition d'une fonction 
def aire_rectangle(longueur, largeur):
    aire = longueur * largeur
    return aire
```

```python
print(aire_rectangle(5,10))
```

```python
print(longueur)
```

```python
print(aire)
```

**Règle 2 :** Vous pouvez utiliser les valeurs des variables globales dans vos fonctions 

```python
# Définition d'une variable globale
pi = 3.14159
```

```python
# Définition d'une fonction qui l'utilise en lecture
def aire_cercle(rayon):
    aire = pi * rayon**2
    return aire
```

```python
# Appel de la fonction 
print(aire_cercle(12))
```

```python
# Affichage variable globale
print(pi)
```

```python
# Affichage variables locales
print(aire)
```

```python
# Affichage variables locales
print(rayon)
```

**Règle 3 :** Vous ne pouvez pas modifier la valeur d'une variable globale dans vos fonctions 

```python
# Définition d'une variable globale
score = 1000
```

```python
# Définition d'une fonction qui l'utilise en écriture
def marquer_points(points):
    score = score + points
```

```python
# Appel de la fonction
marquer_points(50)
```

```python
# Vérification de la variable globale
print(score)
```

**Encore plus important :** Si vous redéfinissez une variable globale dans votre fonction, celle-ci ne sera pas impactée :

```python
# Définition d'une variable globale
score = 5000
```

```python
# Définition d'une fonction utilise la variable en écriture
def reinitialiser_score(points):
    score = points
```

```python
# Appel de la fonction
reinitialiser_score(50)
```

```python
# Vérification de la variable globale
print(score)
```

Alors comment faire ?  

Si vous désirez modifier la valeur d'une variable globale dans votre fonction, vous devez la passer en paramètre à votre fonction et la renvoyer à l'aide d'un `return` :

```python
# Définition d'une fonction qui modifie une variable globale
def marquer_points(sc,points):
    sc = sc + points
    return sc
```

```python
# Appel de la fonction
score = marquer_points(score,50)
```

```python
# Vérification de la variable globale
print(score)
```

## Documenter ses fonctions

### Help
Les fonctions natives de Python comme `print` ou `input` disposent d'un **mode d'emploi**. Pour y accéder, il suffit d'utiliser la commande `help` :

```python
help(print)
```

```python
help(input)
```

!!! tip Aide des fonctions natives de Python
- Lorsque vous utilisez des fonctions, pensez à utiliser la commande `help()`
- Dans le jargon informatique, ce conseil est prodigué avec l'expression [R.T.F.M.](https://fr.wikipedia.org/wiki/RTFM_(expression))
!!!

### les docstrings
Lorsque vous écrivez des fonctions, vous ~~pouvez~~ **devez** créer un mode d'emploi pour celles-ci. En Python, ce mode d'emploi est appelé **docstring**.  
La mise en oeuvre des docstrings est très simple, il suffit d'écrire le mode d'emploi à la suite de la ligne de définition de la fonction (`def`...) et de l'encadrer de trois doubles-guillements `"`. Il est de rigueur de préciser les types de paramètres et des valeurs retournées.  

Il est également fortement conseillé de choisir des **noms évocateurs** pour vos fonctions et variables (cf. cours _Constructions élémentaires._) et de **commenter** cotre code avec des lignes commençant par `#`.

L'exemple ci-dessous, est la réécriture avec docstring, commentaires et noms évocateurs de notre fonction `somme` :

```python
# Définition de la fonction somme avec docstring, commentaires et noms évocateurs

def somme(nb):

    """
    Retourne un entier correspondant à la somme des nb premiers
    entiers naturels. nb est un entier passé en paramètre.
    """
    
    # Initialisation de la somme à 0
    som = 0
    
    # Boucle de 0 à nb
    for entier in range(1, nb + 1):
        # On ajoute chaque entier de la boucle à la somme
        som = som + entier

    # Après la boucle, on renvoie la valeur calculée de la somme
    return som
```

```python
# Appel et affichage du résultat de la fonction
print(somme(500))
```

```python
# Vérification de la docstring
help(somme)
```
