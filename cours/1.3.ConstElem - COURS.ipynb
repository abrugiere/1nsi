## Utilisation des variables

Lorsqu'on modélise un problème avec l'informatique, on voudra manipuler des **données**. Ces données seront amenées à évoluer dans le temps, à **varier**. 

Afin de stocker ces données en mémoire, on peut penser à une boîte sur laquelle est écrit un nom et dans laquelle on place des informations diverses, des valeurs.

Si par exemple on veut stocker le solde de son compte en banque, on pourra utiliser une "boite" qu'on nommera judicieusement `solde` dans laquelle on mettra le montant.

On pourra ensuite dans la modélisation utiliser le mot `solde` à la place de `le solde actuel de mon compte en banque`.

Ce montant est amené à évoluer dans le temps  : augmenter lors de rentrées d'argent (salaire, ventes, virements...) ou diminuer lors de sorties d'argent (achats, prélèvements...)

### Initialisation et affectation

!!! abstract Définitions
- Une variable sert à **stocker** une donnée, elle est définie par son **nom** et contiendra une **valeur** qui peut évoluer dans le temps.
- Lorsqu'on crée une variable en Python pour la première fois, on lui fixe une valeur de départ. On dit que la variable est **initialisée**.
- Lorsqu'ultérieurement on mettra une valeur dans une variable on dit qu'on effectue une **affectation**.
!!!

_Note : En Python, le nom peut être n'importe quelle chaîne alphanumérique, exceptés certains mots réservés, sans commencer par un chiffre._



````python
# Initialisation : la variable a prend la valeur 2
a = 2
print(a)
````

````python
# Initialisation : la variable p prend la valeur 3
b = 3
print(b)
````

````python
# Affectation : la variable d prend la valeur de "2+3" soit "5"
c = 2 + 4
print(c)
````

### Incrémentation et décrémentation

!!! abstract Définition
- Lorsqu'on ajoute 1 à une variable on dit qu'on **incrémente** la variable
- Lorsqu'on enlève 1 à une variable on dit qu'on **décrémente** la variable
!!!


````python
# Création du compteur
compteur = 5
print(compteur)
````

````python
# Incrémentation du compteur
compteur = compteur + 1
print(compteur)
````

Dans ces lignes de codes on fait les étapes suivantes : 
- On évalue l'expression de la partie droite : `compteur + 1` donc `5 + 1` donc `6`  
- On met cette nouvelle valeur dans la variable `compteur`

````python
# Décrémentation du compteur
compteur = compteur - 1
print(compteur)
````

- Lorsqu'on veut mettre incrémenter une variable on peut utiliser la syntaxe suivante :

`x += 1` équivaut à :

```python
x = x + 1
```

_Note : Fonctionne aussi avec `-` `*` `/` et avec d'autres expression que `1`_

````python
# Incrémentation du compteur
nombre = 5
nombre += 1
print(nombre)
````

````python
# Syntaxe avec une multiplication
nb = 4
nb *= 2   # équivaut à nb = nb * 2
print(nb)
````

````python
# Syntaxe avec des variables
p = 10
t = 5
p += t  # équivaut à p = p + t
print(p)
````

### Nommage des variables

- Syntaxiquement, les noms de variables peuvent comporter des **lettres** non accentuées, des **chiffres**, des caractères _ et ne doivent **pas commencer par un chiffre**.

- Quel que soit le langage, on choisira toujours un nom le plus **évocateur** et **explicite** possible, pour l'identifier immédiatement plus tard dans son code.  

Pour donner du sens, il faut souvent plusieurs mots. En Python, pour séparer les mots on utilisera le tiret du bas `_` (underscore, tiret du 8).  
Cette convention d'écriture est appelée le `snake_case`.  

On ne donne jamais un nom de variable au hasard, on le choisit pour qu'il soit explicite.

```python

# BIEN
date == '01-01-2026':
```

## Instructions conditionnelles `if`

- La structure la plus simple est le `if` :
``` python
if condition :
	instructions
```
Le mot **condition**  désigne une expression booléenne (qui vaut soit `True` soit `False`) et le mot **instructions** désigne une instruction ou un bloc d’instructions écrites sur plusieurs lignes à effectuer lorsque la condition est vraie. Exemple :

``` python
if n % 2 == 0 :
	print("Le nombre est pair")
```


- La structure `if-else` ajoute une alternative pour exécuter une autre instruction ou bloc d’instructions à effectuer lorsque la condition est fausse :
``` python
if condition :
	instructions1
else :
	instructions2 
```

Par exemple :
``` python
if n % 2 == 0 :
	print("Le nombre est pair")
else :
	print("Le nombre est impair")
```


- La structure `if-elif-else` ajoute plusieurs alternatives conditionnelles :
``` python
if condition1 :
	instructions1
elif condition2 :
	instructions2
elif condition3 :
	instructions3
...
else :
	instructions4
```

Par exemple :
``` python
if n % 6 == 0 :
	print("Le nombre est multiple de 4")
elif n % 6 == 1 :
	print("Le reste est de 1")
elif n % 6 == 2 :
	print("Le reste est de 2")
else :
	print("Le reste est autre")

```

!!! info Remarques
- Les mots `if` et `elif` sont toujours suivis par une expression qui prendra la valeur `True` ou `False`, la ligne se termine par le symbole `:` (deux points)
- le mot `else` est toujours immédiatement suivi par le symbole `:` (deux points)
- C’est **l’indentation** qui permet de délimiter les blocs d’instructions à exécuter lorsque les conditions sont vérifiées. En Python l'indentation se fait avec la touche de tabulation (`tab`) ou avec 4 espaces.
- Les mots `True`, `False` et `None` sont les rares mots en Python dont l’écriture commence par une lettre majuscule
!!!


````python
# A tester avec 88 et 89
n = 88

if n % 2 == 0 :
    print("Le nombre est pair")
else :
    print("Le nombre est impair")
````

````python
# A tester avec 12 à 17
n = 12

if n % 6 == 0 :
    print("Le nombre est multiple de 4")
elif n % 6 == 1 :
    print("Le reste est de 1")
elif n % 6 == 2 :
    print("Le reste est de 2")
else :
    print("Le reste est autre")
````

## Boucles conditionnelles (`while`)

La structure est la suivante :
``` python
while condition :
	instructions
```

La structure est identique à celle du `if`. Le bloc d’instructions indenté qui suit est exécuté tant que la condition est vérifiée. C’est ce bloc d’instructions qui doit modifier la valeur de la condition pour qu’elle ne soit plus vérifiée et que le programme sorte du bloc et continue son exécution.
``` python
while a  >= b :
    a = a - b
    n = n + 1
```

````python
a = 10
b = 2
n = 0

while a  >= b :
    a = a - b
    n = n + 1

print(a,b,n)
````

### Les dangers de la boucle `While`

- Etudions le programme suivant :

````python
toto = 0
while toto >= 1:
    toto = toto + 1
    print("ce texte ne s'écrira jamais")
print("fini!")
````

Ici la condition `toto >= 1` n'est pas vérifiée dès le départ, on ne rentre donc jamais dans la boucle et on passe directement à la suite

- Etudions maintenant le code suivant :

````python
toto = 1
while toto < 10:
    print("ok")
print("fini!")
````

Cette fois on rentre bien dans la boucle car `toto` vaut 1 et il est donc < 10, mais dans la boucle, on ne le modifie jamais, il restera donc toujours à 1 et on ne sortira jamais de la boucle !

!!! warning Attention
Pour éviter une boucle infinie, il faut **toujours** s'assurer que les instructions de la boucle feront en sorte que la condition ne soit plus vérifiée à un moment.  
_Note :  aucun programmeur professionnel n'est passé à côté de ce piège..._
!!!

## Boucles non conditionnelles  ou itératives (`for`)

Une boucle non conditionnelle sera répétée `n` fois, `n` étant connu à l’avance, la structure est la suivante :
```python
for i in range(n) :
	instructions
```

Ici, une variable `i` est créée et prendra successivement les valeurs `0`,`1`,`2`…,`n-1` et les instructions seront exécutées `n` fois.

````python
for i in range(10):
    print("We're up all night to get lucky")
````

Si nous voulons voir comment évolue la variable `i` affichons-la à chaque **itération** (chaque répétition) :

````python
for i in range(10):
    print(i,"We're up all night to get lucky")
````

Si on ne compte pas utiliser le compteur dans notre code et simplement répéter des instructions, il est possible de remplacer la variable par `_` (underscore ou le tiret du 8). On parlera de **Variable muette** :

````python
for _ in range(10):
    print("We're up all night to get lucky")
````

La fonction `range` ici est utilisée avec un seul argument : le nombre d'itérations, mais il est possible de passer 3 arguments. Si on écrit :
``` python
for k in range(start, end, step):
    print(k)
```

La variable `k` prendra tout d'abord la valeur `start`, progressera de `step` en `step` (le **pas** de la boucle) et s'arrêtera lorsqu'elle sera **strictement** inférieure à `end`.

Par exemple :

````python
for k in range(5,28,3):
    print(k)
````

!!! info Remarques
- `for i in range(f)` : `i` prendra la valeur des entiers successifs de `0` inclus à `f` **exclu**
- `for i in range(d,f)` : `i` prendra la valeur des entiers successifs de `d` inclus à `f` **exclu**
- `for i in range(d,f,p)` : `i` prendra la valeur des de `d` inclus à `f` **exclu avec un pas de** `p`.
!!!

La boucle `for` en Python propose d’autres possibilités, si on considère objet comme une séquence de plusieurs éléments (chaine de caractères, liste…) on pourra utiliser :
``` python
for element in objet :
	instructions
```

Nous reviendrons sur cette possibilité dans le cours sur les tuples et listes.

Exemples :

````python
# Ici la variable car prend successivement les valeurs 'B','o','n','j','o','u','r'
for car in 'Bonjour' :
    print(3 * car)
````

````python
# Ici la variable couleur prend successivement les valeurs 'Vert', 'Jaune', 'Rouge', 'Bleu'
for couleur in ['Vert', 'Jaune', 'Rouge', 'Bleu']:
    print("Un peu de",couleur)
````
