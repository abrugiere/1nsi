## Langages interprétés et compilés


### Ordinateur et langage machine
Pour exécuter un algorithme sur un ordinateur, il faut **traduire** cet algorithme dans une langage que l'ordinateur comprend.  
Le langage du processeur de votre ordinateur est **l'assembleur**. Nous étudierons ce langage de plus près dans le _Chapitre 6 : Architectures matérielles et systèmes d’exploitation_.  
L'assembleur est un langage dit de **bas niveau** c'est à dire qu'il est très proche des instructions natives de votre processeur. Un tel langage est aujourd'hui très peu utilisé car l'écriture d'un code complexe dans ce langage est très fastidieuse.

Cependant, pour faire exécuter les programmes par le processeur de votre ordinateur, il faut toujours **traduire** ces lignes de code en langage machine.  

On peut classer les langages de haut niveau dans deux catégories, issues de la manière dont le code sera traduit en langage machine.


### Langages interprétés

!!! info Définition
Un langage interprété est un langage de programmation dont le code source est traduit puis exécuté ligne par ligne **en temps réel** par l'ordinateur.
!!!
  
Cette transformation du code source en langage machine est effectué par un **interpréteur** ou **machine virtuelle** (_VM_ : Virtual Machine en anglais).  
Pour résumer, lorsque vous écrivez du code dans un langage interprété, c'est ce même code que vous aller directement exécuter sur l'ordinateur quel que soit le système d'exploitation que vous utilisez (Windows, Linux, Mac, Android...).  
Il faudra seulement au préalable avoir installé l'interpréteur qui fonctionne sur votre système d'exploitation.  
Exemples de langages interprétés : Java, Python, Ruby, JavaScript, PHP...


### Langages compilés

!!! info Définition
Un langage compilé est un langage de programmation dont le code source est traduit en langage machine dans un **fichier exécutable**. 
!!!

C'est ce fichier (programme) qui sera exécuté par l'ordinateur.  
Cette transformation du code source en langage machine est effectué par un **compilateur**.  
Comme l'exécutable sera directement exécuté par votre ordinateur, il faudra que celui-ci soit compatible avec votre système d'exploitation, pour un même code source, il faudra donc choisir le bon compilateur afin de générer l'exécutable voulu.  
Les exécutables :
- Sous Windows : `.exe`
- Sous MacOs : `.app`
- Sous Android : `.apk`

Exemples de langages compilés : C, C++, Pascal et OCaml...

### Comparatif des deux types de langages

|             | Avantages    | Inconvénients |
| :---:       |:-------------| :-------------|
| **Compilé**     | - Prêt à l’emploi <br>- Plus performant<br> - Le code source reste privé                    |  - Non multiplateforme <br>- Prise en main et débuggage plus complexe <br>- Étape supplémentaire (compilation) |
| **Interprété**  | - Multiplateforme <br>- Prise en main plus facile et rapide <br>- Plus facile à tester et à débugger  | - Nécessite un interpréteur <br> - Plus lent <br> - Le code source est public |

Les langages interprétés ont pendant longtemps été mal perçus par les développeurs. ## Ecrire ses premières lignes de code Python

### L'instruction `print`

Lorsque on veut afficher un texte avec le langage Python, il faut utiliser l'instruction `print`. Pour cela, il suffit d'écrire le mot clé `print` suivi du texte que l'on souhaite afficher entre parenthèses. Le code suivant affichera `Hello World !` :

```python
>>> print('Hello World !')
```

Vous noterez que la phrase `Hello World !` a été mise entre guillemets, ceci permet de préciser à Python que ceci est bien du texte.

### Résultat d'une opération

La console Python permet, à la manière d'une calculatrice, d'effectuer des opérations :
```python
>>> 12 + 3
```
Renverra `15` (addition)

```python
>>> 10 * 10
```
Renverra `100` (multiplication)

```python
>>> 2 ** 3
```
Renverra `8` (puissance)

### Variable

Il est possible de sauvegarder une valeur dans un élément appelé **variable**, nous verrons ce concept plus en détails dans le prochain cours.

```python
>>> a = 2
```
Stockera la valeur `2` stockée dans l'élément appelé `a`

```python
>>> a
```
Renverra `2` (la valeur stockée dans `a`)

```python
>>> b = 3
```
Stockera la valeur `3` stockée dans l'élément appelé `b`


```python
>>> a + b
```
Renverra la valeur stockée dans l'élément appelé `a` à laquelle elle additionnera stockée dans l'élément appelé `b` donc `2+3` soit `5`


## Éditer et Enregistrer un programme
Comme vous l'avez vu ci-dessus, la console permet d'exécuter des commandes Python **une par une** et lorsque vous allez quitter Python tout ce que vous avez saisi disparaitra.  

La console est donc pratique pour se faire la main ou exécuter des commandes simples, mais ne permet pas l'écriture de programmes complexes qui comporteront plusieurs lignes de code et ne permet pas d'enregistrer son travail.  

### L'éditeur de IDLE
Pour pouvoir écrire des programmes plus longs et les enregistrer, il faudra utiliser l'éditeur d'IDLE, pour le lancer, allez dans le menu de la console Python dans `File` / `New File`, la fenêtre suivante apparaitra :


image

Vous allez pouvoir ici saisir des lignes de code consécutives en allant à la ligne après celles-ci.  
Via le menu `Options` / `Show Line Numbers` vous pourrez voir le numéro des lignes pour faciliter votre travail.

image

### Enregistrer son travail
Avec le menu `File` vous pourrez sauvegarder votre travail (`Save` ou `Save As`).
Le résultat sera un fichier avec l'extension `.py` que vous pourrez ouvrir ultérieurement avec le menu `File` / `Open`

### Exécuter un programme
Maintenant que votre programme est écrit, vous allez pouvoir l'exécuter, pour cela, utilisez le menu `Run` / `Run Module` de l'éditeur ou la touche `F5` du clavier.  
Votre programme sera exécuté et le résultat ce celui-ci sera affiché dans la fenêtre de la console Python.  
Il est donc important d'organiser son espace de travail pour avoir toujours son code et la console côte à côte.

## Quelques fonctions utiles

### Dialogue utilisateur avec l'instruction `input`

Il est possible de demander à l'utilisateur de saisir du texte, puis de récupérer cette saisie afin de l'utiliser ultérieurement.  
Le programme suivant :
- Affiche la phrase `Saisissez du texte : `
- Récupère le texte saisi par l'utilisateur et l'enregistre dans une variable nommée `saisie`, 
- Affiche à l'aide de la fonction `print` la phrase `Vous avez saisi le texte suivant : ` suivie du texte saisi (par l'intermédiaire de la variable) :

```python
saisie = input('Saisissez du texte : ')
print('Vous avez saisi le texte suivant : ',saisie)
```

### Formatage du texte avec le `f-string`
Vous remarquez que la fonction `print` permet d'afficher successivement du texte prédéfini et la valeur d'une variable. Mais il existe une manière encore plus puissante de formater l'affichage : le `f-string`. Cette syntaxte particulière permet d'insérer la valeur de variables directement dans une chaîne de caractères.
Dans l'exemple suivant :

```python
nom = 'Guido van Rossum'
age = 70
print(f'En 2026, le mathématicien {nom} est agé de {age} ans !')
```
- Nous créons une première variable qui contient de texte avec un nom
- Nous créons une deuxième variable qui contient un entier avec un age
- Nous affichons avec `print` une phrase dans laquelle nous insérons directement les valeurs des variables

Et nous obtiendrons : `En 2026, le mathématicien Guido van Rossum est agé de 70 ans !` en retour.

Sans cette fonctionnalité nous aurions été contraints d'écrire le code suivant :
```python
nom = 'Guido van Rossum'
age = 70
print('En 2026, le mathématicien',nom,'est agé de',age,'ans !')
```

### Les commentaires
Dans un programme, il est possible d'écrire du texte qui ne sera pas exécuté par Python, par exemple pour expliquer sa démarche dans un programme, ces lignes sont appelés **commentaires**. Pour commenter une ligne de code, il suffit de la faire commencer par `#` (alt gr + 3).  

!!! info Info
Il est vivement recommandé de mettre des commentaires dans ses programmes afin de le rendre plus compréhensible pour les autres.

## Basthon
L'utilisation d'un IDE nécessite l'installation du logiciel sur son ordinateur.  
Ceci peut être problématique selon le système d'exploitation utilisé ou peut engendrer des problèmes de compatibilité de versions.  
Il existe un outil **en ligne** appelé **basthon** (c'est l'acronyme de "**b**ac **à** **s**able pour py**thon**") qui nous permettra de programmer en Python sans rien installer et quelque soit notre plateforme (PC, mac, Linux, tablette...)  
A la manière de IDLE, il permet de lancer des commandes dans la console Python et de créer, enregistrer et importer des fichiers `.py`  

L'accès au site se fait à l'adresse [https://console.basthon.fr](https://console.basthon.fr)
