# Utilisation de bibliothèques, modularité

## Les bibliothèques

!!! info Définition
Une bibliothèque est une **collection** de plusieurs fichiers (appelés modules) organisés pour répondre à un domaine complet. Elle offre une gamme variée de **fonctionnalités** prêtes à l'emploi pour des projets complexes.
!!!

Lorsque vous utilisez une bibliothèque, vous pourrez disposer de fonctions déjà écrites ou de variables prédéfinies (nous verrons quelques exemples plus tard).


Voici quelques exemples de bibliothèques disponibles en Python que nous utiliserons lors des cours de Première et Terminale NSI :
- **Math** contient des fonctions mathématiques usuelles (racine carrée, arrondi, pi, sinus cosinus) très utile donc pour toutes les applications mathématiques.
- **Random** permet de générer des nombres aléatoires, nous l'utiliserons pour générer des données de tests ou encore faire des tirages au sort pour des statistiques.
- **Matplolib** fournit les outils graphiques pour par exemple tracer des courbes.
- **Turtle** permet à la manière des exercices _Castor_ de début d'année de faire déplacer une tortue dans un environnement graphique.
- **Time** permet de manipuler le temps, utile pour chronométrer des programmes et comparer la performance de différents algorithmes.
- **Tkinter** est la bibliothèque permettant de créer des interfaces graphiques, nous consacrerons un TP à cette bibliothèque.
- **Pandas** regroupe de multiples modules pour manipuler des données, nous l'utiliserons dans le chapitre "Traitement de données en tables"i
- **OS** est une bibliothèque permettant d'intéragir avec le système d'exploitation (gestionnaire de fichiers, exécution de commandes)

!!! info Notes
 - Toutes ces bibliothèques sont disponibles dans l'installation de base de Python (avec IDLE) et dans Basthon Console.
 - Certaines bibliothèques ne sont pas disponibles de base, il faut parfois les installer avec la ligne de commande Windows (ou le terminal Mac ou Linux) et la commande `pip install`. Ceci n'est pas au programme de NSI.
!!!

## Import de bibliothèques
En Python il y a 3 moyens d'importer des fonctions d'une bibliothèque. Regardons un exemple avec la bibliothèque **`math`**, la fonction **`sqrt`** (**sq**uare **r**oo**t** = racine carrée) et la constante **pi** ($\pi$)


### Importer la bibliothèque entière
Ici, toutes les fonctionnalités de la bibliothèque seront disponibles.

```python
# On déclare qu'on va utiliser les fonctions de la bibliothèque math
import math

# On accède aux fonctions en les préfixant du nom de la bibliothèque
print(math.sqrt(81))

# On peut aussi accéder a des constantes disponibles
print(math.pi)
```

### Importer avec un alias

Idem, mais ici l'accès aux fonctions sera plus facile à écrire.


```python
# On déclare qu'on va utiliser les fonctions de la bibliothèque math avec l'alias (raccourci) m
import math as m

# On accède aux fonctions et constantes de la bibliothèque en les préfixant de l'alias
print(m.sqrt(81))
print(m.pi)
```

### Importer une fonction spécifique
Utile si on n'a pas besoin de toutes les fonctionnalités, permet d'alléger son programme. Cependant, cette méthode peut-être dangereuse sur vous définissez des fonctions du même nom que la bibliothèque.

```python
# On déclare qu'on va utiliser la fonction sqrt de la bibliothèque math
from math import sqrt,pi

# On accède à la fonction ou à la consatnte directement
print(sqrt(81))
print(pi)
```

!!! tip Import de bibliothèques
Pour une utilisation simple et sécurisée des bibliothèques, préférez **l'import avec alias**.
!!!

### Accès à la documentation
!!! info docstring
Dans la console Python, on peut accéder à la documentation d'une bibliothèque grace à sa **docstring** et à la commande **`help`** : 
!!!

```python
# On importe la bibliothèque random
import random as rd
```

```python
# Aide sur la bibliothèque complète
help(rd)
```

```python
# Aide sur la fonction randint
help(rd.randint)
```

## La programmation modulaire

Lors du développement d'un logiciel complexe, le nombre de fonctions peut devenir conséquent et le code peut vite faire plusieurs milliers de lignes. Tout concentrer dans un seul fichier Python peut devenir fastidieux à gérer.  
Le principe de modularité consiste à découper son code en plusieurs fichiers appelés **modules** qui contiennent chacun des fonctions regroupées sur un même thème.

!!! info Définition
Un module est un fichier Python `.py` qui contient des fonctions ou constantes regroupées pour une tâche précise.
!!!

Le principe de **modularité** est particulièrement important dans tout développement logiciel. Il simplifie les tests, permet de réutiliser le code et faciliter la maintenance. Ce principe peut opérer à plusieurs niveaux :
- Découper le code en fonctions
- Grouper les fonctions dans un fichier (module)
- Grouper les modules en bibliothèques


La **programmation modulaire** intervient :
- Lors de l'écriture de portions de code pour les réutiliser plus tard ou les distribuer
- Lors de l'utilisation, pour répondre à un besoin, de code écrit par quelqu'un d'autre
- Lors de la répartition à plusieurs de l'écriture du code d'un même programme
