## Algorithmes

!!! info Définition
Un algorithme est suite finie d’étapes ou d'opérations permettant d’obtenir un certain résultat.
!!!

Un peu comme une recette de cuisine, en suivant une liste d’étapes bien précises dans un certain ordre on obtient le même résultat à tous les coups : un délicieux plat.

image

### Historique

Les premiers algorithmes ont été développés bien avant l'émergence de l'informatique.
Les plus anciens algorithmes connus remontent il y a presque quatre millénaires.

- **Vers -1800**, les Babyloniens qui vivaient en Mésopotamie (actuel Irak) utilisaient des algorithmes pour résoudre certaines équations (comme celles du second degré).  
Voici l'image d'une tablette datant de cette période où plusieurs problèmes du second degré sont résolus par une sorte de liste d'instructions proche de nos algorithmes actuels :

image

- **Vers -300**, Euclide a proposé entre autre un algorithme, encore utilisé de nos jours, permettant de déterminer le **P**lus **G**rand **C**ommun **D**iviseur (le **PGCD**) entre deux nombres entiers.  

image

- Le mot algorithme vient du nom latinisé du mathématicien **Al-Khwârizmî**. Ce savant ayant vécu **entre 780 et 850** fut membre de la Maison de la Sagesse de Bagdad. Il répertoria les algorithmes connus à son époque et entre autres travaux il fut l'auteur de deux livres importants :
    - Le premier a conduit au mot « algèbre » actuel ;
    - Le second a permis la diffusion du système de numération décimal actuel à travers le monde : ce sont les « chiffres arabes » actuels.

image

- **En 1842**, Ada Lovelace écrit le premier algorithme de programmation publié de l'histoire. Elle le fait à partir de la machine analytique de Babbage et beaucoup considèrent qu'elle est devenue ainsi la première informaticienne de l'humanité.

image

### Représentation d'un algorithme

La recette est souvent une représentation textuelle ou avec des images, en informatique, un algorithme sera souvent représenté par un schéma comme celui-ci appelé **Organigramme** ou **Logigramme** :

image

Les symboles géométriques courants sont :
- **Ellipse** : indique le début ou la fin de l'algorithme.
- **Rectangle** : représente une action ou un traitement informatique.
- **Parallélogramme** : indique une entrée ou une sortie (lecture ou écriture).
- **Losange** : sert pour les tests conditionnels ou les choix (oui/non).

L'algorithme d'Euclide cité ci-dessus peut-être représenté de la manière suivante :

image

Un algorithme peut être également représenté en langage naturel appelé le **pseudo-code**. Voici la représentation du même algorithme en pseudo-code :    

```
FONCTION PGCD(a, b)
    TANT QUE b != 0 FAIRE
        r <- a modulo b
        a <- b
        b <- r
    FIN TANT QUE
    RETOURNER a
FIN FONCTION

```
La fonction appelée `modulo` ci-dessus correspond au reste de la division euclidienne de `a` par `b`.

## Programmer un ordinateur

Mis à part des calculs (beaucoup et très rapidement), un ordinateur ne sait pas faire énormément de choses... Il sait faire les actions suivantes:
- Gérer des données stockées dans la mémoire
- Comparer des données stockées entre-elles
- Exécuter des actions
- Répéter des actions


### Les instructions
Dans notre recette chaque étape commence par un verbe qui représente les actions à effectuer.
Dans un programme informatique, les étapes sont appelées **instructions** et elles sont également souvent des verbes. 


### Un programme
La révolution de l’informatique survenue au milieu du XXe siècle vient du fait que l’on a pu faire exécuter ces algorithmes à des machines. Pour cela il a fallu exprimer ces algorithmes dans des langages de programmation afin qu’ils puissent être exécutés par les ordinateurs.

!!! info Définition
Un algorithme exécuté par un ordinateur est appelé un programme.
!!!

Programmer, c’est donc exprimer un algorithme dans des langages de programmation afin qu’ils puissent être exécutés par les ordinateurs.  

Il existe différents langages qui permettent de programmer un ordinateur, mais le seul directement utilisable par le processeur est le **langage machine** (suite de 1 et de 0). Aujourd’hui (presque) plus personne ne programme en langage machine (trop compliqué).  


---
## Les langages
Les langages de programmation évolués permettent au programmeur de communiquer avec l'ordinateur dans une langue proche de notre langage naturel.  
Dans les langages de haut niveau, les informaticiens utilisent des instructions (mots souvent en anglais) en lieu et place de la suite de 0 et de 1. Ces instructions, une fois écrites par le programmeur, sont “traduites” en langage machine. 

Il faut donc bien connaître la syntaxe de ces instructions, c'est-à-dire comment les écrire correctement pour que l'ordinateur puisse les interpréter sans erreur (car un ordinateur ne sait pas corriger vos erreurs)
